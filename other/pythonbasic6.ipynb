Why to use object oriented programing and what are benefits of using it
--> Modularity : OOP help in organizing code discrete, reusable units called classes .Each class contains attributes and methods making the code easier to manage,understand and maintain
-->Reusability : With OOP , you can reuse existing code by creating new classes that inherit from existing ones .This inheritance allows for the extensions of functionality without modyfing original code.
--->Maintainability: The structure provide by oops make it easy to update code in large applications by localizing changes to specific parts of the code it becomes easier to modify functionality without affecting entire system
--->readability : IT would make it easier for person to understand the flows and identify the roles of particular modules.


4 Pillars of object oriented programming
--> Encapsulation: its like binding things together.
It allows bundling or wrapping of data(attributes) and methods functioning that operates on data into single units or class.
It hides internal state of the object and all interactions are done at objects methods promoting clear interface and reducing risk of unintended interfernce .
--> Abstraction : Abstraction allows for creating complex system while hiding unnecessary details from user.
It focuses on essential qualities of an object making it simple to interact and understand.
--->Inheritance: Inheritance is a mechanism in oops that allows one class to inherit the attribute and methods of another class.This promotes code reuse anc can lead to heirachial classifications.
---> Polymorphism : It allows object of different class to be treated as object of superclass.It is achieved by method overriding.
In simple terms if we say we can describe it as using same function in different ways .

Dunders / Magic Methods
These are special methods that have __ at begining and __ at end .
These provides a way for a class to define its behaviour
By modifying these methods you can customise the default behaviour of your objects

```python
class Car:
  def __init__(self,name,mileage):
    self.name = name
    self.mileage = mileage
  def __str__(self):
    return f"{self.name}->{self.mileage}"
  def __len__(self):
    return self.mileage
  def __add__(self,other):
    return self.mileage + other.mileage
  def __lt__(self,other):
    return self.mileage < other.mileage
  def __gt__(self,other):
    return self.mileage > other.mileage
  def __eq__(self,other):
    return self.mileage == other.mileage
  def __call__(self):
    print('I was called')

c1 = Car('bmw',20)
c2 = Car('volvo',22)
print(c1)
print(len(c1))
print(c1+c2)
print(c1<c2)
print(c1>c2)
print(c1==c2)
```
