# Dropping Nulls

```python
nulls
```

```python
nulls.dropna(how = 'any', axis = 0) # how = any means any null in the row, axis = 0 means row
```

```python
nulls.dropna(how = 'all', axis = 0) # how = all means all values must be null in the row
```

```python
nulls.dropna(how = 'any', axis = 1) # how = any means any null in the row, axis = 1 means column
```

```python
nulls.dropna(how = 'all', axis = 1) # how = any means any null in the row, axis = 1 means column
```

```python
nulls[nulls['Name'].isnull()].index
```

```python
nulls.drop(index = nulls[nulls['Name'].isnull()].index)
```

```python
nulls.dropna(how = 'all', axis = 1)
```
