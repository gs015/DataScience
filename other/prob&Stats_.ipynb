**Conditional Probability**

```python
--> is the prob. of an event occuring given that another event has already occured
P(A|B) --> prob. of event A given that event B has occured
P(B|A) --> prob. of event B given that event A has occured

P(A|B) = P(A n B)/P(B)

P(A n B) --> P(B|A)*P(A)
```

```python
# Q. In a deck of 52 cards, what is the prob. of drawing a king card given that the
# card drawn is the face card? (face card: Jack, King, Queen)
Solution:
P(K|F) = P(K n F)/P(F)
       = (4/52)/ (12/52)
       = 4/12 = 1/3
```

Q. Suppose a medical test detects a rare disease with a 99% accuracy rate (True postive rate) and 1 % false postive rate. If 0.1% of the population has the disease, what is the prob. that a person has the disease given a positive test result?

```python
A = having disease
B = not having disease
T = test is +ve



P(A|T) = P(T|A)*P(A)/P(T)

-------------------------------------
P(T|A) = 0.99 (prob. of +ve test given the disease)
P(T|-A) = 0.01 (prob. of +ve test given no disease)
P(A) = 1-0.99 = 0.001 (prob. of having disease)
P(B) = 1-0.001 = 0.999 (prob. of not having disease)


P(T)
```
