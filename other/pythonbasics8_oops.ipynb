Getter and Setter method
->It is used when we want to avoid direct access to private variables
->to implement proper encapsulation in python
getter method is used to get access of data members
setter method is used to modify data members

```python
class Person:
  def __init__(self,name,age):
    self.__name=name
    self.__age=age

  #getter method for name
  def get_name(self):
    return self.__name

  #setter method for name
  def set_name(self,name):
    if isinstance(name,str) and name.strip()!="":
      self.__name = name
    else:
      raise ValueError("Invalid name")

  def get_age(self):
    return self.__age

  def set_age(self,age):
    if isinstance(age,int) and age>0:
      self.__age=age
    else:
      raise ValueError("Invalid age")
  def __str__(self):
    return f"Person(name= {self.__name} is of age={self.__age})"

person = Person("Alice",22)


print(person.get_name())
person.get_age()
person.set_age(34)
print(person.get_age())
person.set_name('Tanya')
print(person.get_name())
```

```python
class Encap:
  def __init__(self):
    self.__originalvalue = 100

  def get_value(self):
    return self.__originalvalue

  def set_value(self,value):
    self.__originalvalue = value

  def __str__(self):
    return f"The current value is {self.__originalvalue}"

obj = Encap()
print(obj.get_value())
obj.set_value(2000)
print(obj.get_value())
print(obj)
```

Abstraction
--> hiding irrelevant information

To implement abstraction in python we create an abstract class from abc module
abstract class is a class that cannot be instantiated on its own and designed to be a base class for other classes.
It can contains abstract methods that are declared but contains no implementations

Key points
-> it is used to define common interface for a group od related classes
->Its object cannot be created
->Its methods is not directly implemented



```python
from abc import ABC,abstractmethod
class Sample(ABC):
  @abstractmethod
  def display(self):
    print("This is main sample class ")
class Child(Sample):
  def display(self):
    super().display()
    print("This is an abstract method")
    print("Hello")

obj = Child()
obj.display()
```

Decorators --> Decorators are powerful tool in python that allows you to modify or enhance the behaviour of your function without permanently modifying their definition

A decorator is  a function that takes another function as an argument ,add some functionality and returns new function
It is used using '@' sign.

```python
def my_decorator(func):
  def wrapper():
    print("Hello we have arrived")
    func()
    print("Hello we are about to leave now ")
  return wrapper

@my_decorator
def say_hello():
  print("Please welcome have dinner with us ")

say_hello()
```

```python
def my_decorator(func):
  def wrapper(*args,**kwargs):
    print(f"calling function {func.__name__} with args {args} and kwargs {kwargs}")
    result = func(*args,**kwargs)
    print(f"returned result is {result}")
    return result
  return wrapper

@my_decorator
def add(x,y):
  return x+y

@my_decorator
def multiply(x,y,z):
  return x*y*z

add(10,20)
multiply(100,4,7)
```

```python
def my_login(func):
  def wrapper():
    print("login is done or not please check")
    func()
    print("login is done")
  return wrapper

@my_login
def add_to_cart():
  print("add to cart")

add_to_cart()
```

Method overriding :
-> Run time Polymorphism
-> occurs at run time based on actual type of object.
->Involves base class and one or more derived class with overriden methods.
->decision of which method to call is made at run time based on actual type of object.

```python
class Animal:
  def speak(self):
    return "Animal speaks"
class Dog:
  def speak(self):
    return "Dog barks"

#Method overriding
animal = Animal()
dog = Dog()

animal_sound = animal.speak()
print(animal_sound)
dog_sound = dog.speak()
print(dog_sound)
```

Method Overloading 🇱-
-> Compile-time(static) Polymorphism
->It occurs at compile time based on number or types of argument passed to the function.
->decision of which method to call is made at compile time based on provided arguments.


```python
class Calc:
  def add(self,a,b,c=0):
    return a+b+c

  def display(self,a,b):
    return self.add(a,b)



class Calc1:
  def add(self,a,b):
    return a+b

c1 = Calc()
c2 = Calc1()
print(c2.add(10,20))
print(c1.add(10,20,50))
#print(c1.display(10,20,30))
print(c1.display(10,20))
```

```python
def add(datatype,*args):
  if datatype =='int':
    result = 0
    for arg in args:
      result += arg
    return result
  elif datatype == 'float':
    result = 0.0
    for arg in args:
      result += arg
    return result

  if datatype =='str':
    result = ''
    for x in args:
      result = result + x

    return result

print(add('int',20,30))
print(add('float',20,90.0))
print(add('str','Python','Programming'))
```
