```python
def bub_sort(array):
  for i in range(0, len(array)):
    for j in range(0, len(array)- i - 1):
      if array[j] < array[j + 1]:
        temp = array[j]
        array[j] = array[j + 1]
        array[j+1] = temp
  return array

my_array = np.array([20,14,25,16,45,60,12,9])
bub_sort(my_array)

```

```python
# The behavior of np.around in NumPy with rounding halfway values (like 2.5 and 3.5) can be surprising at first. Here's the reason:

# np.around uses Banker's rounding which rounds a number to the nearest even integer when it's exactly halfway between two integers.

# 2.5: In this case, 2 and 3 are the nearest integers. Since 2.5 is exactly halfway, Banker's rounding kicks in and rounds it down to the nearest even integer, which is 2.
# 3.5: Here, the nearest integers are 3 and 4. Again, 3.5 sits exactly in the middle. However, since the fractional part (0.5) is even, Banker's rounding rounds it up to the nearest even integer, which is 4.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
mu = 0     # mean
sigma = 1  # standard deviation

# Generate data
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, mu, sigma)

# Plot
plt.plot(x, y)
plt.title("Standard Normal Distribution")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.grid()
plt.show()
```

```python
import numpy as np
import scipy.stats as stats

# Sample data
sample = np.array([78, 65, 20, 69, 36, 81, 85, 71, 44, 71, 47, 37, 89, 25, 73, 55, 80, 52, 46, 77])

# Basic stats
n = len(sample)
mean = np.mean(sample)
std = np.std(sample, ddof=1)  # sample std dev
confidence = 0.95
df = n - 1

# t critical value
t_crit = stats.t.ppf((1 + confidence) / 2, df)

# Margin of error
margin_error = t_crit * (std / np.sqrt(n))

# Confidence Interval
lower = mean - margin_error
upper = mean + margin_error

(lower, upper)
print(t_crit, mean)
```
