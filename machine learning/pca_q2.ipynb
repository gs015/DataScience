```python
ss_gym = StandardScaler().fit_transform(df_gym.iloc[:,:-1])
```

```python
cov_gym = np.cov(ss_gym.T)
cov_gym
```

```python
eig_val, eig_vec = np.linalg.eig(cov_gym)
```

```python
eig_val
```

```python
eig_vec
```

```python
# cumulative sum of explained variance ratio in percentage
tot_var = eig_val.sum()
explained_var_ratio = [v / tot_var * 100 for v in sorted(eig_val, reverse=True)]
explained_var_ratio
```

```python
tot_var
```

```python
np.cumsum(explained_var_ratio)
```
