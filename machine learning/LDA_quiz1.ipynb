```python
X = pd.DataFrame(data=digit_dict.data, columns=digit_dict.feature_names)
```

```python
# y = pd.Series()
y = digit_dict.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=19)
```

```python
ss_model = StandardScaler()
x_train_scaled = ss_model.fit_transform(x_train)
x_test_scaled = ss_model.transform(x_test)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
lda_model = LinearDiscriminantAnalysis(n_components=9)
```

```python
x_train_lda = lda_model.fit_transform(x_train_scaled, y_train)
x_test_lda = lda_model.transform(x_test_scaled)
```

```python
x_train_lda.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_model = LogisticRegression()
```

```python
lr_model.fit(x_train_lda, y_train)
```

```python
y_pred2 = lr_model.predict(x_test_lda)
```

```python
accuracy_score(y_test, y_pred2)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=28)
```

```python
rf_model.fit(x_train_lda, y_train)
```

```python
y_pred = rf_model.predict(x_test_lda)
```

```python
accuracy_score(y_test, y_pred)
```

```python
y_test_series = pd.Series(y_test)
y_pred_series = pd.Series(y_pred)
```

```python
y_test_series
```

```python
# (y_pred_series[y_test_series==3]==3).sum() / (y_test_series==3).sum()
pd.DataFrame(classification_report(y_test, y_pred, output_dict=True))
```

```python
recall_score(y_test, y_pred, average=None)
```
