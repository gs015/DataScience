K-Means Clustering

```python
df2 = pd.read_csv('/content/OnlineRetail.csv', encoding='latin-1')
df2.info()
```

```python
df2 = df2.dropna(subset=['CustomerID'])
df2['CustomerID'] = df2['CustomerID'].astype('str').str.replace('.0', '')
```

```python
df2.info()
```

```python
# add plot
```

```python
# customer segmentation
df2 = df2[df2['Quantity'] > 0]
df2['TotalPrice'] = df2['Quantity'] * df2['UnitPrice']
df2['InvoiceDate'] = pd.to_datetime(df2['InvoiceDate'])
```

```python
# RFM
snapshot_date = df2['InvoiceDate'].max() + dt.timedelta(days=1)

rfm = df2.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'InvoiceNo': 'count',
    'TotalPrice': 'sum'
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']
```

```python
rfm
```

**K-Means Clustering**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

rfm_scaled = StandardScaler().fit_transform(rfm)

kmeans_model = KMeans(n_clusters=4, random_state=42)
rfm['Cluster'] = kmeans_model.fit_predict(rfm_scaled)
```
