```python
import numpy as np
import matplotlib.pyplot as plt
```

**Objective:**


The primary objective is to predict the output value y based on two input features
x1 and x2 using a basic neural network model with the equation y=3x1+4x2. The
neural network will be trained using a feed-forward and back-propagation
mechanism to minimize the error between the predicted and actual values.

**Activation Functions**

* Define the sigmoid function, a popular activation function used in neural networks.

* Define the tanh function, which is another common activation function that returns values between -1 and 1.

* Define the relu function, which returns positive values as they are and zeroes out negative values.



```python
# f(x) = 1/1+exp(x)
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
#relu : f(x) = max(0,x)
def relu(x):
  return np.max(0, x)
```

```python
# Tanh (hyperbolic tangent) is a type of activation function
# that transforms its input into a value between -1 and 1.
def tanh(x):
  # return np.tanh(x)
  return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))
```

2. Sample Data:

* Use the following sample data for training: x1=3, x2=2, yactual=17

3. Feed Forward:
* Initialize the random weights w1 and w2 between 1 and 10.
* Predict the output ypred using the equation ypred= x1 x w1 + x2 x w2

```python
x1 = 3
x2 = 2
y = 3 * x1 + 4 * x2 #17
```

```python
# learning rate and random weights
lr = 0.01
w1 = np.random.randint(1, 11)
w2 = np.random.randint(1, 11)
error = np.array([])
y_pred = np.array([])
```

```python
for e in range(25):
  y_pred = np.append(y_pred, np.array([w1 * x1 + w2 * x2])) #feed forward
  error = np.append(error, np.array((y_pred[e] - y) ** 2))
  # back probagation - derivatives of weights and update
  derviative_w1 = 2 * (y_pred[e] - y) * x1
  derivative_w2 =  2 * (y_pred[e] - y) * x2
  w1 = w1 - (lr * derviative_w1)
  w2 = w2 - (lr * derivative_w2)
  # print(w1,w2,y_pred[e])
```

```python
print(f'Predictions: {y_pred}')
print(f'Error: {error}')
```

```python
plt.figure(figsize=(5,5))
plt.xlabel('Epochs')
plt.ylabel('Error')
plt.title('Epochs vs Error')
plt.plot(error)
plt.show()
```
