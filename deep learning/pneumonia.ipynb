```python
IMAGE_SIZE = 224
```

A generator function : is a special type of function that **returns an iterator object**. Instead of using return to send back a single value, generator functions use yield to produce a series of results over time.

The function **pauses** its execution **after** **yield**, maintaining its **state** between iterations.

it helps Memory Efficient , Pipeline Processing. https://www.geeksforgeeks.org/python/generators-in-python/


```python
# Rewrite using python's generator (yield), set image size, and convert numpy array
# python generator to read images
def image_generator(fpath, batch_size=32):
  path = os.path.join(data_path, fpath)
  # print(data_path, fpath, path)
  x = list()
  try:
    for i, fn in enumerate(os.listdir(path), start=1):
      # print(os.path.join(path, fn))
      file_array = cv2.resize(cv2.imread(os.path.join(path, fn)),
                                (IMAGE_SIZE, IMAGE_SIZE)) / 255
      # print(file_array)
      x.append(file_array)
      if i % batch_size == 0:
        # print(i, x)
        yield np.array(x) # in generator object the x has been maintained and its values kept in memory there.

        #can also vanish x like below:
        # if len(x_batch) == batch_size:
        #     yield np.array(x_batch), np.array(y_batch)
        #     x_batch, y_batch = [], [] # next iteration starts here

  except Exception as e:
    print('Exception :' + str(e))
  return x
```

```python
x_train = image_generator(fpath='train/PNEUMONIA', batch_size=42)
# x_train = image_generator('/train/PNEUMONIA')
# x_train = [*x_train]
```
